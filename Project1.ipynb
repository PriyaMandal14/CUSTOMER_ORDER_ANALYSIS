# 1. Store Customer Order

```python
# List of customer names
customers = ["Priya", "Raja", "Aashu", "Divs"]

# List of order details as tuples: (customer_name, product, price, category)
orders = [
    ("Priya", "Laptop", 900, "Electronics"),
    ("Raja", "T-shirt", 20, "Clothing"),
    ("Aashu", "Table", 50, "Home Essentials"),
    ("Priya", "Headphones", 100, "Electronics"),
    ("Divs", "Dress", 80, "Clothing"),
    ("Raja", "Microwave", 100, "Home Essentials"),
    ("Aashu", "Smartphone", 600, "Electronics"),
    ("Divs", "Blender", 150, "Home Essentials")
]

# Dictionary mapping customer names to their orders
customer_orders = {}
for name in customers:
    customer_orders[name] = []

for order in orders:
    customer_orders[order[0]].append(order[1:])
```

#2. Classify products by category

```python
# Map products to categories
product_category_map = {order[1]: order[3] for order in orders}

# Set of unique product categories
unique_categories = set(product_category_map.values())

# Display all available product categories
print("Available Product Categories:", unique_categories)
```

# 3. Analyze customer orders

```python
# Classify customers by spending
customer_spending = {}
customer_classification = {}

for customer, items in customer_orders.items():
    total = sum([item[1] for item in items])
    customer_spending[customer] = total
    if total > 100:
        customer_classification[customer] = "High-Value Buyer"
    elif 50 <= total <= 100:
        customer_classification[customer] = "Moderate Buyer"
    else:
        customer_classification[customer] = "Low-Value Buyer"
```

# 4. Generate business insights

```python
# Total revenue per category
category_revenue = {}
for order in orders:
    category = order[3]
    category_revenue[category] = category_revenue.get(category, 0) + order[2]

# Unique products
unique_products = set([order[1] for order in orders])

# Customers who purchased electronics
electronics_customers = [order[0] for order in orders if order[3] == "Electronics"]

# Top 3 high-spending customers
top_spenders = sorted(customer_spending.items(), key=lambda x: x[1], reverse=True)[:3]
```

# 5. Organize and display data

```python
# Summary of each customer’s total spending and their classification
for customer in customers:
    print(f"{customer}: ${customer_spending[customer]} - {customer_classification[customer]}")

# Customers who purchased from multiple categories
customer_categories = {name: set() for name in customers}
for order in orders:
    customer_categories[order[0]].add(order[3])

multi_category_customers = [name for name, cats in customer_categories.items() if len(cats) > 1]

# Customers who bought both electronics and clothing
electronics_buyers = set([order[0] for order in orders if order[3] == "Electronics"])
clothing_buyers = set([order[0] for order in orders if order[3] == "Clothing"])
common_buyers = electronics_buyers & clothing_buyers
```
